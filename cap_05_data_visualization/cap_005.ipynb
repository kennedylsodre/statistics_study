```python
from numpy import sin
import matplotlib.pyplot as plt 
from random import seed,randint
```

```python
#Plotting a sine wave
x_axis =[x*0.1 for x in range(100 )] 
y = sin(x_axis)

plt.plot(x_axis, y, label='sin(x)')
```

```python
seed(1)  
x_axis = ['green','red','blue'] 
y_axis = [randint(0,100),randint(0,100),randint(0,100)]
plt.bar(x_axis, y_axis, label='Random Values', color=x_axis)
```
